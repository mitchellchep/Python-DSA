```python
#calculating the number of seconds in a day
days = int(input("Enter number of days"))
def seconds_per_day(days):
    #introduce the variables, how many seconds are in a day, step by step conversion
    hours = days*24
    minutes = hours*60
    seconds = minutes*60
    return seconds
print(f"There are {seconds_per_day(days)} in {days}  days.")
```

```python
#program that calculates the volume of a sphere 
import math    
radius = float(input("Input radius: "))
def volume_of_a_sphere(radius):
    volume =(4/3)*math.pi*radius**3
    return volume
print(f"The volume of the sphere is {volume_of_a_sphere(radius)}")
```

```python
 # compute the area and perimeter of a square
#since its a square the values are the same
values = int(input("Input the value of your square"))
def area_of_square(values):
    area = values**2
    return area
def perimeter_of_square(values):
    perimeter = values*4
    return perimeter
print(f"The area and perimeter of your square are {area_of_square(values)} and {perimeter_of_square(values)} respectively.")
```

```python
# checks for upper or lower case in characters

yourCharacter = input("Input a character:")
def check(yourCharacter):
    if(yourCharacter >= 'A' and yourCharacter <= 'Z'):
        print(f"{yourCharacter} is an uppercase character");
    elif(yourCharacter >= 'a' and yourCharacter <= 'z'):
        print(f"{yourCharacter} is a lowercase character");
    else:
        print(f"{yourCharacter} is not an alphabetic character");
check(yourCharacter)
```

```python
numbers = []#initialize an emptylist to store the array
print("Please enter 5 numbers:")
#continually prompt  a user for 5 values
for i in range (5):
    val = float(input("Enter value {i + 1}:")) #converts to float
    numbers.append(val)#populating the array
average = sum(numbers)/len(numbers) #average by dividing the sum by array lenght
print("\nValues entered into the array:", numbers)
print(f"The average of the values is: {average}")
    
```
